# Problem Statement

**Restaurants and individuals can list surplus food.**

**NGOs or individuals in need can claim the food.**

**SQL stores available food details and locations.**

**A Streamlit app enables interaction, filtering, CRUD operation and visualization.**

**BUSINESS USE CASE:**

**Connecting surplus food providers to those in need through a structured platform.**

**Reducing food waste by redistributing excess food efficiently.**

**Enhancing accessibility via geolocation features to locate food easily.**

**Data analysis on food wastage trends for better decision-making.**

The three status categories — Pending, Cancelled, and Completed — are almost in equal proportions, at around 33% each.

Claim stays usually uniform, with peaks around March 12 and March 18, both exceeding 60 total claims.

Pending claims form a large share of the total, while Cancelled claims fluctuate more sharply.

Completed claims remain consistent, showing less variation than other statuses.

```python
# Quantity of food claimed over time

df['Timestamp']= pd.to_datetime(df['Timestamp'])

# Group by date
df_daily = (
    df.groupby(df["Timestamp"].dt.date)["Quantity"]
    .sum()
    .reset_index()
)

# Line chart
fig = px.line(
    df_daily,
    x="Timestamp",
    y="Quantity",
    title="Quantity of Food Claimed Over Time"
)
fig.show()
```

Food claims fluctuate daily, with noticeable spikes around mid-month.

There’s a steady decline after March 18, with the lowest claims observed on March 21, which may indicate reduced supply or less food availability.

The pattern suggests a cyclical demand, where claims rise every few days before dropping, possibly reflecting weekly distribution schedules or donation cycles.

```python
# Expiry date proximity vs claims

df['Expiry_Date']= pd.to_datetime(df['Expiry_Date'])
df['Timestamp']= pd.to_datetime(df['Timestamp'])

df['Days_to_Expiry'] = (df['Expiry_Date'] - df['Timestamp']).dt.days

```

```python
# quantity claimed by expiry time range

import numpy as np

# Separate expired and non-expired
expired_df = df[df["Days_to_Expiry"] < 0]
non_expired_df = df[df["Days_to_Expiry"] >= 0]

# Create bins
max_days = int(np.ceil(non_expired_df["Days_to_Expiry"].max()))
bins = range(0, max_days + 5, 5)
labels = [f"{bins[i]}-{bins[i+1]} days" for i in range(len(bins)-1)]

non_expired_df["Expiry_Range"] = pd.cut(
    non_expired_df["Days_to_Expiry"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Group and sum quantities
non_expired_bins = non_expired_df.groupby("Expiry_Range")["Quantity"].sum().reset_index()

expired_sum = expired_df["Quantity"].sum()
expired_row = pd.DataFrame({"Expiry_Range": ["Expired"], "Quantity": [expired_sum]})

df_bins = pd.concat([expired_row, non_expired_bins], ignore_index=True)

# Plot
fig = px.bar(
    df_bins,
    x="Expiry_Range",
    y="Quantity",
    title="Quantity Claimed by Expiry Time Ranges (Including Expired)",
    text="Quantity"
)
fig.show()
```

The highest claimed quantities are for items expiring within 10–15 days (6,314 units), followed closely by 5–10 days and 0–5 days ranges.

Expired items still account for 1,433 units, indicating room for improvement in timely distribution.

Claims drop sharply beyond 15 days to expiry, suggesting lower urgency or demand for longer shelf-life items.
